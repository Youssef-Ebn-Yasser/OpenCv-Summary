```python
import cv2
import datetime
```

```python
cap = cv2.VideoCapture(0)  # 0  from device or 'path/of/video'
```

```python
fourcc = cv2.VideoWriter_fourcc(*'XVID')
```

```python
out = cv2.VideoWriter('output.avi' ,fourcc , 30.0 ,(640,480))  
#(vedioName ,fourcc=hasManyTypes [enhance vedios], numberOfFramesPerSecond , (width , height)  )
```

```python
while(cap.isOpened()) :
    ret , frame = cap.read()  # boolean ret 
    if ret == True :
        print(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        print(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

        ## show current date in image
        text = 'Width' + str(cap.get(3)) + 'Height' + str(cap.get(4)) 
        date = str(datetime.datetime.now)
        frame = cv2.putText(frame,text , (50, 450),cv2.FONT_HERSHEY_SIMPLEX,1.5,(255, 255, 255),3)
        frame = cv2.putText(frame,date , (50, 450),cv2.FONT_HERSHEY_SIMPLEX,1.5,(255, 255, 255),3)

        
        # change default window size
        cap.set(3,720)  # cv2.CAP_PROP_FRAME_WIDTH == 3
        cap.set(4,720)  # cv2.CAP_PROP_FRAME_HEIGHT == 4
        out.write(frame)

        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        cv2.imshow('frame' , gray)

        if cv2.waitKey(1)  & 0xFF == ord('q'):
            break

        else :
            break

# cap.release()
# out.release()
# cv2.destroyAllWindows()
```

```python
"""
cv2.EVENT_FLAG_ALTKEY  =>  Alt key is pressed down.
cv2.EVENT_FLAG_CTRLKEY => Control key is pressed down.
cv2.EVENT_FLAG_SHIFTKEY =>	Shift key is pressed down.
cv2.EVENT_FLAG_LBUTTON => Left mouse button is active/held down.
cv2.EVENT_FLAG_RBUTTON => Right mouse button is active/held down.
cv2.EVENT_FLAG_MBUTTON => Middle mouse button is active/held down.

cv2.EVENT_MOUSEMOVE	=> Triggered when the mouse cursor is moved inside the window.
cv2.EVENT_MOUSEWHEEL =>	Vertical mouse wheel is scrolled.
cv2.EVENT_MOUSEHWHEEL => Horizontal mouse wheel is scrolled (if supported).
    
cv2.EVENT_LBUTTONDOWN	Left mouse button is pressed down.
cv2.EVENT_RBUTTONDOWN	Right mouse button is pressed down.
cv2.EVENT_MBUTTONDOWN	Middle mouse button (wheel click) is pressed down.
    
cv2.EVENT_LBUTTONUP	Left mouse button is released.
cv2.EVENT_RBUTTONUP	Right mouse button is released.
cv2.EVENT_MBUTTONUP	Middle mouse button is released.
    
cv2.EVENT_LBUTTONDBLCLK	Left mouse button is double-clicked.
cv2.EVENT_RBUTTONDBLCLK	Right mouse button is double-clicked.
"""
```

```python
import cv2
import numpy as np

# 1. Define the Mouse Callback Function
def interactive_painter(event, x, y, flags, param):
    # Retrieve extra data passed via 'userdata' (param unpacks to canvas color settings)
    drawing_settings = param 
    canvas = drawing_settings["canvas"]
    
    # Event: Left Button Down -> Draw a Blue Circle
    if event == cv2.EVENT_LBUTTONDOWN:
        cv2.circle(canvas, (x, y), drawing_settings["size"], (255, 0, 0), -1)
        print(f"Left click at ({x}, {y})")

    # Event: Right Button Down -> Draw a Green Square
    elif event == cv2.EVENT_RBUTTONDOWN:
        size = drawing_settings["size"]
        cv2.rectangle(canvas, (x - size, y - size), (x + size, y + size), (0, 255, 0), -1)
        print(f"Right click at ({x}, {y})")

    # Event: Mouse Wheel -> Dynamically adjust shape size
    elif event == cv2.EVENT_MOUSEWHEEL:
        delta = cv2.getMouseWheelDelta(flags)
        if delta > 0:
            drawing_settings["size"] = min(100, drawing_settings["size"] + 2)
        else:
            drawing_settings["size"] = max(5, drawing_settings["size"] - 2)
        print(f"Wheel scrolled. New brush size: {drawing_settings['size']}")

def main():
    # Create a blank black image canvas (512x512 pixels, 3 channels for BGR color)
    img = np.zeros((512, 512, 3), dtype=np.uint8)
    
    # Dictionary container to pass as userdata (avoids global scope reliance)
    user_data = {
        "canvas": img,
        "size": 15  # Initial radius/half-width for shapes
    }

    window_name = "OpenCV Mouse Event Demo"
    cv2.namedWindow(window_name)

    # 2. Register the Mouse Callback to the window and pass user_data
    cv2.setMouseCallback(window_name, interactive_painter, user_data)

    print("Instructions:")
    print(" - Left Click: Draw blue circle")
    print(" - Right Click: Draw green square")
    print(" - Scroll Wheel: Change brush size")
    print(" - Press 'q' to exit.")

    while True:
        cv2.imshow(window_name, user_data["canvas"])
        
        # Wait for 20ms and check if 'q' key is pressed to break loop
        if cv2.waitKey(20) & 0xFF == ord('q'):
            break

    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```
